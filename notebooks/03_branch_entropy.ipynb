# Branch Entropy Analysis

This notebook analyzes the evolution of Shannon entropy in quantum amplitude distributions.


```python
import sys
sys.path.append('../src')

from predictive_agent import PredictiveAgent
from quantum_branching import AmplitudeBrancher
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Entropy evolution experiment
actions = ["attack", "block", "idle"]
agent = PredictiveAgent(actions, seed=0, n_hypotheses=4)

entropies = []
observations = ["attack", "block", "attack", "idle", "block", "attack"]

for obs in observations:
    # Record current entropy
    entropies.append(agent.brancher.entropy())
    
    # Update with observation
    agent.update_with_feedback(obs)
    
    print(f"Observed: {obs}, Entropy: {agent.brancher.entropy():.3f}")

# Plot entropy evolution
plt.figure(figsize=(10, 6))
plt.plot(entropies, 'bo-')
plt.xlabel('Observation Step')
plt.ylabel('Shannon Entropy')
plt.title('Entropy Evolution with Observations')
plt.grid(True)
plt.show()
```
