# Qiskit Mapping for Quantum Hardware

This notebook demonstrates how to map amplitude distributions to quantum circuits using Qiskit.

**Note**: Requires `pip install qiskit` for quantum hardware experiments.


```python
# Uncomment to install qiskit
# !pip install qiskit

try:
    from qiskit import QuantumCircuit, transpile
    from qiskit.quantum_info import Statevector
    QISKIT_AVAILABLE = True
except ImportError:
    QISKIT_AVAILABLE = False
    print("Qiskit not available. Install with: pip install qiskit")

import sys
sys.path.append('../src')
from quantum_branching import AmplitudeBrancher
import numpy as np
```

```python
if QISKIT_AVAILABLE:
    # Create amplitude brancher
    brancher = AmplitudeBrancher(n_hypotheses=4)
    
    # Apply some operations
    brancher.apply_unitary_like_mix(mix=0.1)
    brancher.collapse_with_observation([0.8, 0.2, 0.1, 0.9])
    
    print("Amplitude distribution:", brancher.amplitudes)
    print("Probabilities:", brancher.amplitudes**2)
    print("Entropy:", brancher.entropy())
    
    # Map to quantum circuit (simplified example)
    n_qubits = 2  # For 4 hypotheses
    qc = QuantumCircuit(n_qubits)
    
    # Create superposition state
    qc.h(0)
    qc.h(1)
    
    print("\\nQuantum circuit created for hardware mapping")
    print(qc.draw())
else:
    print("Qiskit not available - skipping quantum circuit creation")
```
