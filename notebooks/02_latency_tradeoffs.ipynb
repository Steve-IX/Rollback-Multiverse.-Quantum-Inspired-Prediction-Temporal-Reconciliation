# Latency Tradeoffs Analysis

This notebook analyzes the tradeoffs between network latency and rollback frequency.


```python
import sys
sys.path.append('../src')

from environments import OneHPDuel
from rollback_engine import RollbackReconciler
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Latency sweep experiment
latencies = [10, 30, 50, 100, 200]
rollback_rates = []

for latency in latencies:
    env = OneHPDuel(seed=42)
    rollback_engine = RollbackReconciler()
    
    # Run simulation
    for _ in range(500):
        obs, r0, r1, info = rollback_engine.simulate_frame(
            env, "attack", "block",
            latency_ms=(latency, latency),
            jitter_ms=(5, 5)
        )
    
    stats = rollback_engine.get_stats()
    rollback_rates.append(stats['rollback_rate'])
    print(f"Latency {latency}ms: Rollback rate = {stats['rollback_rate']:.3f}")

# Plot results
plt.figure(figsize=(10, 6))
plt.plot(latencies, rollback_rates, 'bo-')
plt.xlabel('Latency (ms)')
plt.ylabel('Rollback Rate')
plt.title('Rollback Rate vs Network Latency')
plt.grid(True)
plt.show()
```
