# Introduction to Rollback Multiverse Simulation

This notebook introduces the basic concepts and components of the rollback multiverse system.


```python
import sys
sys.path.append('../src')

from environments import OneHPDuel
from predictive_agent import PredictiveAgent
from rollback_engine import RollbackReconciler
import numpy as np
```

## Basic Environment Setup


```python
# Initialize environment
env = OneHPDuel(seed=42)
obs = env.reset()
print("Initial observation:", obs)
```

## Predictive Agent with Quantum Branching


```python
# Initialize predictive agent
agent = PredictiveAgent(env.ACTIONS, seed=0, n_hypotheses=4)

# Make a prediction
predicted_action, probs = agent.predict_opponent()
print(f"Predicted opponent action: {predicted_action}")
print(f"Hypothesis probabilities: {probs}")
print(f"Current entropy: {agent.brancher.entropy():.3f}")
```

## Rollback Engine Simulation


```python
# Initialize rollback engine
rollback_engine = RollbackReconciler()

# Simulate a few frames
for i in range(5):
    obs, r0, r1, info = rollback_engine.simulate_frame(
        env, "attack", "block", 
        latency_ms=(30, 30), 
        jitter_ms=(5, 5)
    )
    print(f"Frame {i}: Rollback={info['rollback']}, Rewards=({r0}, {r1})")

print("\nRollback stats:", rollback_engine.get_stats())
```
